# Local Government Revenue
Cities, counties and special districts report finances to the State Controller's Office (SCO). Fiscal years 2002-03 on.

**Read this first**
- **Not included:** K-12 school districts and community colleges report to the Department of Education, not the SCO, so they are not here. Most property tax ultimately funds schools, so these charts understate total local tax collection. County and city tax charts below show what cities and counties themselves receive.
- **Double counting:** counties receive large state and federal aid ("intergovernmental"), which also shows up in state spending. Do not add state and local totals without netting this out.
- **Reporting break:** SCO changed its category format in FY2016-17, so some groups shift at the dotted line. Internal service funds and fund-type categories (special districts) only exist after the change.
- Values are as reported by each government (SCO does not audit them), in nominal dollars unless the toggle says otherwise.

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from ca_taxes.io import PROCESSED
from ca_taxes.deflate import add_views, base_fy, LAST_ACTUAL_FY
from ca_taxes.style import stacked_area_toggle, base_layout, PALETTE, SOURCE_NOTE
pio.renderers.default = "plotly_mimetype"
SCO_NOTE = "Source: California State Controller's Office, Cities / Counties / Special Districts Annual Reports"
BREAK = {2016.5: "SCO reporting format changes"}
lr = pd.read_csv(PROCESSED / "local_revenue.csv")
for govt in ["City", "County", "Special districts"]:
    d = add_views(lr[lr.govt == govt].groupby(["fy", "group"], as_index=False).amount_k.sum())
    stacked_area_toggle(d, "group", f"{govt} revenue by source", SCO_NOTE, breaks=BREAK).show()
```
