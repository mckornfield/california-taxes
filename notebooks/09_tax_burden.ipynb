# Total Taxes per Person
State taxes plus city and county taxes, per Californian. This counts taxes California governments *collect*, not what households pay (federal taxes and school-district levies are excluded, and some state tax is paid by businesses and visitors).

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from ca_taxes.io import PROCESSED
from ca_taxes.deflate import add_views, base_fy, LAST_ACTUAL_FY
from ca_taxes.style import stacked_area_toggle, base_layout, PALETTE, SOURCE_NOTE
pio.renderers.default = "plotly_mimetype"
SCO_NOTE = "State = all funds (LAO / DOF). County and city = SCO. Excludes school and special districts."
st = pd.read_csv(PROCESSED / "state_revenue.csv")
lt = pd.read_csv(PROCESSED / "local_taxes.csv")
a = st[st.group == "Tax"].groupby("fy").amount_k.sum().rename("State")
b = lt[lt.govt == "City"].groupby("fy").amount_k.sum().rename("City")
c = lt[lt.govt == "County"].groupby("fy").amount_k.sum().rename("County")
d = pd.concat([a, b, c], axis=1).dropna().reset_index().melt("fy", var_name="level", value_name="amount_k")
d = add_views(d)
d = d[d.fy <= 2024]
stacked_area_toggle(d, "level", "Taxes collected: state, county and city", SCO_NOTE, order=["State", "County", "City"]).show()
```
